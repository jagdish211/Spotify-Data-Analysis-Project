## DATA CLEANING Artists Data Farme

```python
# Artists Data Farme

# Filling Missing Value in country column 
artists['country'] = artists['country'].fillna('Unknown')

# Filling Missing Value in debut_year column 
artists['debut_year'] = artists['debut_year'].fillna(
    artists['debut_year'].median()
)
```

```python
# cheking Null Value 
print(artists.isnull().sum())
```

```python
# Exporting cleanded file 
artists.to_csv(r"C:\Users\rampa\DS Python Class\Spotify-Data-Analysis\data\cleaned\artists_clean.csv", index=False)
```
